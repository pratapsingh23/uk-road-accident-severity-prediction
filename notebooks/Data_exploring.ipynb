# Insights

In this dataset, missing values are stored as `-1`, not as NaN.

### casualty_df
1. Here are only three `sex_of_casualty`, `age_of_casualty`, and `casualty_type` columns contain missings values.
2. Missing values are very few. So, I can replace by `np.nan`

### vehicle_df
1. Some columns have very few missing values.
2. Some columns have up to about 40% missing values, so I will replace `-1` with `np.nan`.

### collision_df
1. Many columns have very few missing values.
2. Some columns have up to about 40% missing values, so I will replace `-1` with `np.nan`.
3. `junction_control` have 42.41% missing values. So, I can drop those rows or either I can replace with `np.nan`

**Replacing -1 with np.nan**

```python
# Replace -1 (missing data marker) with np.nan in casualty_df
casualty_df = casualty_df.replace(-1, np.nan)

# Replace -1 with np.nan in vehicle_df
vehicle_df = vehicle_df.replace(-1, np.nan)

# Replace -1 with np.nan in collision_df
# junction_control:
display(pd.crosstab(collision_df['junction_detail'], collision_df['junction_control']==-1))

mask = (collision_df['junction_control'] == -1) & (collision_df['junction_detail'] == 0)
collision_df.loc[mask, 'junction_control'] = 0

collision_df = collision_df.replace(-1, np.nan)
```

# Summary of Notebook

1. I used three datasets for this project: casualty, vehicle and collision.
2. The datasets come from the official UK government website (link in the README.md file of the repo).
3. Shape of the raw datasets (before column selection):

6. I replaced the remaining `-1` values with `np.nan` in all three datasets.
