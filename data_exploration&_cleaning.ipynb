```python
# dropping unwanted columns:
collision_raw = collision_df.copy()

collision_cols = [
    'collision_index', 'collision_severity', 'date', 'day_of_week',
    'time', 'latitude', 'longitude', 'urban_or_rural_area', 'speed_limit', 'road_type', 'light_conditions', 'weather_conditions', 'road_surface_conditions', 'special_conditions_at_site', 'carriageway_hazards', 'junction_detail', 'junction_control', 'trunk_road_flag', 'number_of_vehicles', 'number_of_casualties', 'did_police_officer_attend_scene_of_accident'
]

collision_df = collision_df[collision_cols]
```

##**Meaning of each selected column:**

collision_index: Unique identifier code assigned to each individual accident record.

collision_severity: Classification of the accident's severity level (Fatal, Serious, or Slight).

date: The exact calendar date on which the collision occurred.

day_of_week: The day of the week when the accident took place.

time: The precise time of day when the collision occurred.

latitude / longitude: Geographical coordinates marking the exact physical location of the accident.

urban_or_rural_area: Classification of the collision site as an urban or rural environment.

speed_limit: The maximum legal speed limit applicable to the road segment.

road_type: The structural layout of the road (e.g., Single carriageway, Dual carriageway, Roundabout).

light_conditions: The illumination level at the scene (e.g., Daylight, Darkness with street lights).

weather_conditions: Atmospheric weather state at the time of impact (e.g., Fine, Rain, Fog, Snow).

road_surface_conditions: The physical state of the road surface (e.g., Dry, Wet, Ice, Snow).

special_conditions_at_site: Exceptional operational hazards present at the site (e.g., roadworks, defective traffic signals).

carriageway_hazards: Physical obstructions or dangers present on the road (e.g., fallen load, animal in carriageway).

junction_detail: The specific physical configuration of any nearby junction (e.g., T-junction, Crossroads).

junction_control: The traffic management mechanism present at the junction (e.g., Give way, Traffic lights, Stop sign).

trunk_road_flag: Binary flag indicating whether the collision occurred on a major national trunk road.

number_of_vehicles: Total count of all vehicles involved in the accident.

number_of_casualties: Total count of individuals injured or killed in the collision.

did_police_officer_attend_scene_of_accident: Indicator showing whether a police officer visited and recorded the scene.

#**2. Casualty df, exploring and cleaning**

```python
casualty_df.head()
```

```python
casualty_df.columns
```

```python
casualty_raw = casualty_df.copy()

casualty_cols = [
    'collision_index','casualty_type', 'escooter_flag', 'pedestrian_movement',
    'casualty_imd_decile'
]

casualty_df = casualty_df[casualty_cols]
```

```python
casualty_df.info()
```

```python
casualty_df.head()
```

```python
casualty_df['casualty_type'] = casualty_df['casualty_type'].replace([-1, 99], np.nan)
casualty_df['pedestrian_movement'] = casualty_df['pedestrian_movement'].replace([9, -1], np.nan)
casualty_df['casualty_imd_decile'] = casualty_df['casualty_imd_decile'].replace(-1, np.nan)
print("NUll values")
casualty_df.isnull().sum()
```

```python
casualty_df['escooter_flag'].value_counts()
```

```python
# casualty aggregation
tmp = casualty_df[['collision_index', 'casualty_imd_decile']].copy()
tmp['has_vulnerable_casualty'] = casualty_df['casualty_type'].isin([0,1, 2, 3, 4, 5, 16, 22, 23, 97])
tmp['has_escooter_casualty'] = (casualty_df['escooter_flag'] == 1).astype(int)
tmp['has_pedestrian_movement'] = casualty_df['pedestrian_movement'].between(1,8).astype(int)

cas_agg = tmp.groupby('collision_index').agg(
    has_vulnerable_casualty = ('has_vulnerable_casualty', 'max'),
    has_escooter_casualty=('has_escooter_casualty', 'max'),
    has_pedestrian_movement=('has_pedestrian_movement', 'max'),
    avg_casualty_imd=('casualty_imd_decile', 'mean')
).reset_index()


print(cas_agg)
```

#**3. Vehicle df, exploring and cleaning**
