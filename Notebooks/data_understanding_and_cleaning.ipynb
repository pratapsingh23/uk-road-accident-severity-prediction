```python
# Importing required libaries:
import pandas as pd
import numpy as np
```

```python
# Opening the CSV file:
df = pd.read_csv('/content/drive/MyDrive/accident_data.csv')
df.head(10)
```

```python
# Checking Rows and columns:
print(f"Rows: {df.shape[0]} \nColumns: {df.shape[1]}")
```

```python
# Making a copy of dataset:
df_clean = df.copy()
```

```python
# Checking column names:
print(df_clean.columns)

# Converting into simple and usable forms:
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.replace(' ', '_')
    .str.lower()
)

# ---FIXED---
print("\n----FIXED----\n")
print(df_clean.columns)
```

```python
# Checking information of dataset
df_clean.info()
```

```python
# Changing 'accident_date' Dtype from 'object' to 'date time'

df_clean['accident_date'] = pd.to_datetime(df_clean['accident_date'], format="%d-%m-%Y")

```

```python
# Checking Null values percentages :
null_value_pct = df_clean.isnull().mean() * 100
print(null_value_pct[null_value_pct > 0])
```

# Handling Null Values

1. Some columns, such as `latitude`, `longitude` and `urban_or_rural_area`, contain
   only a few null values (fewer than 30 rows in total). Since this is a tiny fraction
   of the dataset, I will drop those rows, as doing so will not affect the analysis.

2. Other columns, such as `road_surface_conditions`, `road_type` and
   `weather_conditions`, contain a larger number of null values. I will not drop
   these rows, as that would mean losing useful data. Instead, I will fill the
   missing values with a separate category called "Unknown".

```python
# Dropping rows:
df_clean = df_clean.dropna(subset=['latitude', 'longitude', 'urban_or_rural_area'])

# making seprate category:
for col in ['road_surface_conditions', 'road_type', 'weather_conditions']:
    df_clean[col] = df_clean[col].fillna('Unknown')

df_clean.isnull().sum()
```

```python
# Checking duplicates:
print(f"Duplicates rows count{df_clean.duplicated().sum()}")

print("\n----Duplicates rows----\n")
dups = df_clean[df_clean.duplicated(keep=False)]
print(dups.sort_values(list(df_clean.columns)).head(20))
```

# Handling Duplicate Values

```python
# Dropping duplicates:
df_clean = df_clean.drop_duplicates()
```

# Latiitude/Longitude Validity checking:

In UK,
Latitude is generally vary  49.5 to 61.
Longitude is genrally vary -8 to 2

So, I need to check and varify latitude/longitude is real in dataset?

```python
# Latitude and longitude validity:

# -- How data looks?
print(df_clean[['latitude', 'longitude']].describe())

# -- Checking zero values:
print("\nLatitude: 0 =>", (df_clean['latitude'] == 0).sum())
print("\nLongitude: 0 =>", (df_clean['longitude'] == 0).sum())

# -- Counting rows outside from UK range:
valid_mask = (
    df_clean['latitude'].between(49, 61) &
    df_clean['longitude'].between(-9, 2)
)

valid_rows = df_clean[valid_mask]
invalid_rows = df_clean[~valid_mask]

print("\nValid rows:", len(valid_rows))
print("\nInvalid rows:", len(invalid_rows))

"""
In this dataset, there are no invlalid rows available.
So, I skip it now and move to next
"""
```
